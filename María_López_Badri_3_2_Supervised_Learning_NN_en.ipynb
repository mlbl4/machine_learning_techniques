## Neighborhood-based Techniques

- These techniques are based on distance functions that allow comparing input samples pairwise to obtain a measure of similarity or resemblance.
- One of the most used algorithms is called k-NN (k Nearest Neighbors), which relies on the target values of the nearest neighbors to:
   - estimate the density function and calculate the posterior probability, $P(ω_i |x) = k_i /k$, that a sample $x$ belongs to class $ω_i$ ($k_i$ is the number of neighbors of class i) in classification problems;
   - estimate the expected value in regression problems, usually by calculating the mean or median of the target values of the nearest neighbors.

## Distance Function

In this case, we will use the most common one, the `Euclidean` or `L2` distance.

$$d_{L2}(\overrightarrow{x},\overrightarrow{y})=\sqrt{\sum_i(x_i-y_i)^2}$$

where $x$ and $y$ are vectors of numbers.

````python
def d_L2(x,y):
  # Compute the Euclídea or L2 distance between two numeric vectors
  return np.sqrt(np.sum((x-y)**2))

# Test case
x = np.array([1,2])
y = np.array([2,3])

# Result 1.41
d_L2(x,y)
````

## Example 1: Calculation of the Nearest Neighbor

We are going to implement a function whose parameters are a series of items and a set so that it calculates the nearest neighbor of each item to the given set.


````python
def nearest_neighbors(sample, X_train):
  '''
    Find the nearest neighbor of a new sample to a training set (X_train)
  '''
  min_distance = float('inf')
  nearest_neighbor = None

  for x in X_train:
    distance = d_L2(sample, x)
    if distance < min_distance:
      min_distance = distance
      nearest_neighbor = x

  return  [sample, nearest_neighbor]

# Example test
X_test = np.array([ [1, 2], [1, 1], [8, 8] ])
X_train = np.array([ [1, 2], [3, 4], [5, 6], [7, 8] ])

# for each position in x_test array we see which one is the nearest form the X_train
for sample in X_test:
  print(nearest_neighbors(sample, X_train))
````

## Example 2: Classification using Nearest Neighbors

Implement a Python function whose parameters are a set of samples with their respective classes (labels) and a sample to classify using the nearest neighbors algorithm. In this case, you can use the mode (the most repeated label) of the nearest neighbors as the label assigned to the new sample.


````python
import statistics

def k_nearest_neighbors(X_train, y_train, sample, k, debug=False):
  '''
    Find the most common k labels among the nearest neighbors
  '''

  distances = []

  for x, y_label in zip(X_train,y_train):
      distance = d_L2(sample, x)
      distances.append((distance, y_label))

  # TODO: Sort all distances and their labels in ascending order
  distances.sort()
  if debug:
    print(f'sample: {sample} all distances: {distances}')

  # TODO: Get the most repeated label for the first k elements.
  labels = [label for _, label in distances[:k]]
  most_repeated_label = statistics.mode(labels) # Add code

  return most_repeated_label

# Use case
X_train = np.array([ [1, 2], [3, 4], [5, 6], [7, 8] ])
y_train  = ['A', 'B', 'A', 'B']
X_test  = np.array([[1,2], [4, 5]])
k = 3

for sample in X_test:
  print(k_nearest_neighbors(X_train, y_train, sample, k, debug=True))
````
